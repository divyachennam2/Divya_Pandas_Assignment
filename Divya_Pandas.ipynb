### 3. Explain How NumPy Broadcasting Enables the Addition of These Arrays

NumPy broadcasting allows arrays with compatible shapes to be added element-wise. Here, the array `a` with shape `(3, 1)` is expanded across the columns, while `b` with shape `(1, 3)` is expanded across the rows. Therefore, both arrays are effectively treated as shape `(3, 3)`.

The addition becomes:

```text
[[10+1, 10+2, 10+3],
 [20+1, 20+2, 20+3],
 [30+1, 30+2, 30+3]]
 ```
 Hence, the result has shape (3, 3).

match according to NumPy's broadcasting rules.

Reason: NumPy cannot broadcast an array of shape (2,) with an array of shape (3, 1).

### Q1. A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing. 



### 1. Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().


````python
import pandas as pd

df = pd.read_csv("student_assessment_dirty.csv", keep_default_na=False)

print("First 5 rows:")
print(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
````

### 2. Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.


````python
df["Department"] = df["Department"].str.strip().str.upper()

df["Department"] = df["Department"].replace({
    "I.T.": "IT"
})

print(df["Department"].unique())
````

### 3. Identify and remove exact duplicate student assessment records

````python
duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)

df = df.drop_duplicates()

print("Records after removing duplicates:", len(df))
````

### 4. Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.


````python
df["Marks_Status"] = "Entered"

df.loc[df["Marks"].str.strip() == "Absent", "Marks_Status"] = "Absent"

df.loc[
    df["Marks"].str.strip().isin(["NA", ""]),
    "Marks_Status"
] = "Not Entered"

df["Marks"] = df["Marks"].str.extract(r"(\d+(?:\.\d+)?)")[0]

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df[["Student_ID", "Marks", "Marks_Status"]])
````

### 5. Check misisng values in the cleaned DataFrame 

````python
print(df[df["Marks"].isna()][
    ["Student_ID", "Student_Name", "Marks_Status"]
])
````

##### Explanation: This identifies records where marks are missing and distinguishes students who were Absent from students whose marks were Not Entered.

##### Risk: Deleting missing values without investigating why they are missing may remove valid information, such as an absent student's record, and can affect the accuracy of the analysis.

### 6. Identify Marks Outside the Valid Range of 0–100


````python
invalid_marks = df[
    (df["Marks"] < 0) | (df["Marks"] > 100)
]

print("Invalid marks:")
display(invalid_marks[["Student_ID", "Student_Name", "Marks"]])
````

##### Explanation: Marks below 0 or above 100 are identified as invalid. In the given dataset, S114 has 105 marks, so this value should be investigated before using it in analysis.

### 7. Display the Cleaned DataFrame and Report Results

````python
missing_marks_remaining = df["Marks"].isna().sum()

invalid_marks_identified = len(invalid_marks)

print("Cleaned DataFrame:")
print(df)

print("\nRecords removed:", duplicate_count)
print("Missing marks remaining:", missing_marks_remaining)
print("Invalid marks identified:", invalid_marks_identified)
````
